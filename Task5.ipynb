# Logistical Regression


```python
features = [
    "Fuel Consumption City (L/100 km)",
    "Fuel Consumption Hwy (L/100 km)",
    "Fuel Consumption Comb (L/100 km)",
    "Fuel Consumption Comb (mpg)"
]

target = "CO2 Emissions(g/km)"

data = df[features + [target]].copy()

median_co2 = data[target].median()

data["high_emissions"] = (
    data[target] > median_co2
).astype(int)

print("Median CO2 emissions:", median_co2)
print(data[features + ["high_emissions"]].head())
```

```python
X = data[features].to_numpy(dtype=float)
y = data["high_emissions"].to_numpy(dtype=float)

print("Input shape:", X.shape)
print("Target shape:", y.shape)
```

```python


indices = np.random.permutation(len(y))
split = int(0.8 * len(y))

train_idx = indices[:split]
test_idx = indices[split:]

X_train = X[train_idx]
X_test = X[test_idx]

y_train = y[train_idx]
y_test = y[test_idx]
```

```python
mean_X = X_train.mean(axis=0)
std_X = X_train.std(axis=0)

# Avoid division by zero for any constant feature
std_X[std_X == 0] = 1

X_train_scaled = (X_train - mean_X) / std_X
X_test_scaled = (X_test - mean_X) / std_X
```

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))
    
```

```python
# Start with zero coefficients
b0 = 0.0
weights = np.zeros(X_train_scaled.shape[1])

learning_rate = 0.1
epochs = 2000
n = len(y_train)

for i in range(epochs):

    # 1. Calculate the score using all four features
    z = b0 + X_train_scaled @ weights

    # 2. Convert scores into probabilities
    probabilities = sigmoid(z)

    # 3. Calculate errors
    errors = probabilities - y_train

    # 4. Calculate gradients
    db0 = np.mean(errors)
    dw = (X_train_scaled.T @ errors) / n

    # 5. Update the coefficients
    b0 -= learning_rate * db0
    weights -= learning_rate * dw

print("Intercept:", b0)

for feature, weight in zip(features, weights):
    print(feature, "coefficient:", weight)
```

```python
z = b0 + X_train_scaled @ weights
```

```python
z_test = b0 + X_test_scaled @ weights

probabilities = sigmoid(z_test)

predictions = (probabilities >= 0.5).astype(int)

print("Actual labels:   ", y_test[:10].astype(int))
print("Predicted labels:", predictions[:10])
```

```python
accuracy = np.mean(predictions == y_test)

print("Accuracy:", round(accuracy * 100, 2), "%")
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    np.arange(len(y_test)),
    probabilities,
    c=y_test,
    cmap="coolwarm",
    alpha=0.7
)

plt.axhline(
    0.5,
    color="black",
    linestyle="--",
    
    label="Decision threshold"
)

plt.xlabel("Test vehicle index")
plt.ylabel("Predicted probability of high emissions")
plt.title("Logistic Regression: Predicted Probabilities")
plt.legend()
plt.grid(True)
plt.show()
```
